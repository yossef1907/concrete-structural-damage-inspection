```python
import zipfile

zip_path = '/content/archive (4) (1).zip'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall('/content/crack_data')
```

```python
import os

print(os.listdir('/content/crack_data'))
```

```python
dataset_train = tf.keras.utils.image_dataset_from_directory(
    '/content/crack_data',
    validation_split=0.2,
    subset='training',
    seed=42,
    image_size=(224, 224),
    batch_size=32,
    label_mode='binary'
)
```

```python
dataset_validation = tf.keras.utils.image_dataset_from_directory(
    '/content/crack_data',
    validation_split=0.2,
    subset='validation',
    seed=42,
    image_size=(224, 224),
    batch_size=32,
    label_mode='binary'
)
```

```python
plt.figure(figsize=(10,5))

for i, (image, label) in enumerate(dataset_train.take(1)):
    for j in range(6):
        plt.subplot(2,3,j+1)
        plt.imshow(image[j].numpy().astype("uint8"))
        plt.title("Negative" if label[j].numpy() == 0 else "Positive")
        plt.axis("off")

plt.show()
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./255, input_shape=(224, 224, 3)),
    tf.keras.layers.Conv2D(32, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Conv2D(128, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    dataset_train,
    validation_data=dataset_validation,
    epochs=10
)
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(['Train', 'Validation'])

plt.subplot(1,2,2)
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(['Train', 'Validation'])

plt.show()
```

```python
loss, accuracy = model.evaluate(dataset_validation)

print("Validation Loss:", loss)
print("Validation Accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

y_true = []
y_pred = []

for images, labels in dataset_validation:
    predictions = model.predict(images, verbose=0)
    y_true.extend(labels.numpy().astype(int).flatten())
    y_pred.extend((predictions >= 0.5).astype(int).flatten())

cm = confusion_matrix(y_true, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Negative', 'Positive']
)

disp.plot()
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_true,
    y_pred,
    target_names=['Negative', 'Positive']
))
```

```python
model.save('/content/crack_detection_model.keras')
print("Model saved successfully")
```
