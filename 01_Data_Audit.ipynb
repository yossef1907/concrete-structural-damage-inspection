```python
# %% [markdown]
# # 01 — Data Audit
# **Concrete Damage Detection & Segmentation Project**
# 
# Objective: Perform a complete audit of the available concrete damage datasets before training.

# %% 1. Import Libraries
import os
from pathlib import Path
from collections import Counter, defaultdict
import hashlib
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import display

print("Libraries imported successfully.")

# %% 2. Dataset Path
DATA_ROOT = Path(r"D:\nti_project\data")

print("Dataset root:")
print(DATA_ROOT)
print("\nExists:", DATA_ROOT.exists())

# %% 3. Define Supported Image Extensions
IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff",
    ".webp"
}

print("Supported image extensions:")
print(IMAGE_EXTENSIONS)

# %% 4. Show Dataset Structure
def print_tree(path, max_depth=4, current_depth=0):
    path = Path(path)
    if current_depth > max_depth:
        return

    items = sorted(path.iterdir(), key=lambda x: (x.is_file(), x.name.lower()))

    for item in items:
        indent = "    " * current_depth
        if item.is_dir():
            print(f"{indent}📁 {item.name}/")
            print_tree(item, max_depth, current_depth + 1)
        else:
            print(f"{indent}📄 {item.name}")

if DATA_ROOT.exists():
    print_tree(DATA_ROOT, max_depth=4)
else:
    print(f"Path does not exist: {DATA_ROOT}")

# %% 5. Find All Images
def get_image_files(folder):
    folder = Path(folder)
    return [
        p for p in folder.rglob("*")
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

all_images = get_image_files(DATA_ROOT) if DATA_ROOT.exists() else []
print(f"Total images found: {len(all_images):,}")

# %% 6. Basic Image Statistics
print("First 20 images:\n")
for image_path in all_images[:20]:
    print(image_path)

# %% 7. Analyze Folders
folder_stats = []

if DATA_ROOT.exists():
    for folder in sorted(DATA_ROOT.rglob("*")):
        if folder.is_dir():
            images = [
                p for p in folder.iterdir()
                if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
            ]
            if len(images) > 0:
                folder_stats.append({
                    "Folder": str(folder.relative_to(DATA_ROOT)),
                    "Image Count": len(images)
                })

folder_df = pd.DataFrame(folder_stats)
if not folder_df.empty:
    folder_df = folder_df.sort_values(by="Image Count", ascending=False).reset_index(drop=True)
    display(folder_df)

# %% 8. Dataset Size Summary
total_dataset_images = folder_df['Image Count'].sum() if not folder_df.empty else 0
print(f"Total number of images: {len(all_images):,}")
print(f"Number of folders containing images: {len(folder_df):,}")
print(f"Total dataset size: {total_dataset_images:,}")

# %% 9. Analyze File Extensions
extension_counts = Counter(image.suffix.lower() for image in all_images)
extension_df = pd.DataFrame(
    extension_counts.items(),
    columns=["Extension", "Count"]
).sort_values(by="Count", ascending=False)

display(extension_df)

# %% 10. Plot File Extensions
if not extension_df.empty:
    plt.figure(figsize=(8, 5))
    plt.bar(extension_df["Extension"], extension_df["Count"])
    plt.xlabel("File Extension")
    plt.ylabel("Number of Images")
    plt.title("Image File Extensions")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# %% 11. Check Corrupted Images
corrupted_images = []

for image_path in all_images:
    try:
        with Image.open(image_path) as img:
            img.verify()
    except Exception as e:
        corrupted_images.append({
            "Path": str(image_path),
            "Error": str(e)
        })

corrupted_df = pd.DataFrame(corrupted_images)
print(f"Corrupted images: {len(corrupted_df)}")

if len(corrupted_df) > 0:
    display(corrupted_df.head(20))
else:
    print("No corrupted images found.")

# %% 12. Analyze Image Dimensions
dimension_records = []

for image_path in all_images:
    try:
        with Image.open(image_path) as img:
            width, height = img.size
            dimension_records.append({
                "Path": str(image_path),
                "Width": width,
                "Height": height,
                "Channels": len(img.getbands()),
                "Format": img.format,
                "Mode": img.mode
            })
    except Exception:
        pass

dimensions_df = pd.DataFrame(dimension_records)
if not dimensions_df.empty:
    display(dimensions_df.head())

# %% 13. Dimension Statistics
if not dimensions_df.empty:
    print("Width statistics:")
    display(dimensions_df["Width"].describe())
    print("\nHeight statistics:")
    display(dimensions_df["Height"].describe())

# %% 14. Most Common Image Dimensions
if not dimensions_df.empty:
    dimension_counts = (
        dimensions_df
        .groupby(["Width", "Height"])
        .size()
        .reset_index(name="Count")
        .sort_values("Count", ascending=False)
    )
    display(dimension_counts.head(20))

# %% 15. Image Mode Analysis
if not dimensions_df.empty:
    mode_counts = dimensions_df["Mode"].value_counts().reset_index()
    mode_counts.columns = ["Mode", "Count"]
    display(mode_counts)

# %% 16. Find Non-RGB Images
if not dimensions_df.empty:
    non_rgb = dimensions_df[dimensions_df["Mode"] != "RGB"]
    print(f"Non-RGB images: {len(non_rgb)}")
    display(non_rgb.head(20))

# %% 17. Duplicate Detection Helper
def calculate_md5(file_path, chunk_size=8192):
    md5 = hashlib.md5()
    with open(file_path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            md5.update(chunk)
    return md5.hexdigest()

# %% 18. Calculate Image Hashes
hash_records = []
for image_path in all_images:
    try:
        file_hash = calculate_md5(image_path)
        hash_records.append({
            "Path": str(image_path),
            "Hash": file_hash
        })
    except Exception:
        pass

hash_df = pd.DataFrame(hash_records)
print(f"Images hashed: {len(hash_df):,}")

# %% 19. Find Exact Duplicates
if not hash_df.empty:
    duplicate_groups = (
        hash_df
        .groupby("Hash")
        .filter(lambda x: len(x) > 1)
        .sort_values("Hash")
    )
    print(f"Images belonging to duplicate groups: {len(duplicate_groups):,}")
    display(duplicate_groups.head(50))
else:
    duplicate_groups = pd.DataFrame()

# %% 20. Number of Unique Images
if not hash_df.empty:
    unique_images = hash_df["Hash"].nunique()
    total_hashed = len(hash_df)
    duplicate_count = total_hashed - unique_images
else:
    unique_images, total_hashed, duplicate_count = 0, 0, 0

print(f"Total images: {total_hashed:,}")
print(f"Unique images: {unique_images:,}")
print(f"Duplicate copies: {duplicate_count:,}")

# %% 21. Create Dataset-Level Image Table
records = []
for image_path in all_images:
    relative_path = image_path.relative_to(DATA_ROOT)
    try:
        with Image.open(image_path) as img:
            width, height = img.size
            records.append({
                "Path": str(relative_path),
                "Folder": str(relative_path.parent),
                "Filename": image_path.name,
                "Extension": image_path.suffix.lower(),
                "Width": width,
                "Height": height,
                "Mode": img.mode
            })
    except Exception:
        records.append({
            "Path": str(relative_path),
            "Folder": str(relative_path.parent),
            "Filename": image_path.name,
            "Extension": image_path.suffix.lower(),
            "Width": None,
            "Height": None,
            "Mode": None
        })

images_df = pd.DataFrame(records)
if not images_df.empty:
    display(images_df.head())

# %% 22. Save Image Inventory
OUTPUT_DIR = DATA_ROOT.parent / "audit_results"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

if not images_df.empty:
    images_df.to_csv(OUTPUT_DIR / "image_inventory.csv", index=False)
    print("Saved:", OUTPUT_DIR / "image_inventory.csv")

# %% 23. Analyze Main Classification Folders
classification_folders = ["Crack", "Background", "Negative", "No crack", "Positive"]
classification_stats = []

for folder_name in classification_folders:
    folder_path = DATA_ROOT / folder_name
    if folder_path.exists():
        images = get_image_files(folder_path)
        classification_stats.append({"Class": folder_name, "Images": len(images)})
    else:
        classification_stats.append({"Class": folder_name, "Images": 0})

classification_df = pd.DataFrame(classification_stats)
display(classification_df)

# %% 24. Classification Distribution Plot
if not classification_df.empty:
    plt.figure(figsize=(9, 5))
    plt.bar(classification_df["Class"], classification_df["Images"])
    plt.xlabel("Folder")
    plt.ylabel("Number of Images")
    plt.title("Classification Dataset Distribution")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

# %% 25 & 26. Show Classification Samples
def show_samples(folder_path, n=6):
    images = get_image_files(folder_path)
    if len(images) == 0:
        print("No images found.")
        return

    samples = random.sample(images, min(n, len(images)))
    plt.figure(figsize=(15, 8))

    for i, image_path in enumerate(samples):
        try:
            image = Image.open(image_path)
            plt.subplot(2, 3, i + 1)
            plt.imshow(image)
            plt.title(image_path.name)
            plt.axis("off")
        except Exception:
            pass

    plt.tight_layout()
    plt.show()

for folder_name in classification_folders:
    folder_path = DATA_ROOT / folder_name
    if folder_path.exists():
        print(f"\n===== {folder_name} =====")
        show_samples(folder_path, n=6)

# %% 27 - 31. Damage Detection Dataset Inspection
DETECTION_ROOT = DATA_ROOT / "Damage Detection"
DETECTION_IMAGES = DETECTION_ROOT / "Images"
YOLO_LABEL_DIR = DETECTION_ROOT / "Labels" / "Yolo"
VOC_LABEL_DIR = DETECTION_ROOT / "Labels" / "Pascal VOC"

detection_images = get_image_files(DETECTION_IMAGES) if DETECTION_IMAGES.exists() else []
yolo_labels = list(YOLO_LABEL_DIR.rglob("*.txt")) if YOLO_LABEL_DIR.exists() else []
voc_labels = list(VOC_LABEL_DIR.rglob("*.xml")) if VOC_LABEL_DIR.exists() else []

detection_summary = pd.DataFrame({
    "Item": ["Images", "YOLO Labels", "Pascal VOC Labels"],
    "Count": [len(detection_images), len(yolo_labels), len(voc_labels)]
})
display(detection_summary)

# %% 32. Check Detection Image-Label Matching
def get_stem_set(files):
    return {file.stem for file in files}

image_stems = get_stem_set(detection_images)
yolo_stems = get_stem_set(yolo_labels)

missing_yolo_labels = image_stems - yolo_stems
extra_yolo_labels = yolo_stems - image_stems

print(f"Images without YOLO labels: {len(missing_yolo_labels)}")
print(f"YOLO labels without images: {len(extra_yolo_labels)}")

# %% 33 - 35. Damage Segmentation Dataset Analysis
SEGMENTATION_ROOT = DATA_ROOT / "Damage Segmentaion"
difficulty_levels = ["Easy", "Medium", "Hard"]
segmentation_stats = []

for level in difficulty_levels:
    level_path = SEGMENTATION_ROOT / level
    image_dir = level_path / "Images"
    mask_dir = level_path / "Labels" / "Mask"

    images = get_image_files(image_dir) if image_dir.exists() else []
    masks = get_image_files(mask_dir) if mask_dir.exists() else []

    segmentation_stats.append({
        "Difficulty": level,
        "Images": len(images),
        "Masks": len(masks)
    })

segmentation_df = pd.DataFrame(segmentation_stats)
display(segmentation_df)

# %% 36 - 39. Show Segmentation Image + Mask
def show_segmentation_samples(level, n=4):
    level_path = SEGMENTATION_ROOT / level
    image_dir = level_path / "Images"
    mask_dir = level_path / "Labels" / "Mask"

    images = get_image_files(image_dir)
    if len(images) == 0:
        print("No images found.")
        return

    samples = random.sample(images, min(n, len(images)))
    plt.figure(figsize=(12, 3 * len(samples)))

    for i, image_path in enumerate(samples):
        mask_path = None
        for extension in IMAGE_EXTENSIONS:
            candidate = mask_dir / f"{image_path.stem}{extension}"
            if candidate.exists():
                mask_path = candidate
                break

        plt.subplot(len(samples), 2, 2 * i + 1)
        image = Image.open(image_path)
        plt.imshow(image)
        plt.title(f"{level} - Image")
        plt.axis("off")

        plt.subplot(len(samples), 2, 2 * i + 2)
        if mask_path is not None:
            mask = Image.open(mask_path)
            plt.imshow(mask, cmap="gray")
            plt.title("Mask")
        else:
            plt.title("Mask not found")
        plt.axis("off")

    plt.tight_layout()
    plt.show()

for level in difficulty_levels:
    if (SEGMENTATION_ROOT / level).exists():
        show_segmentation_samples(level, n=4)

# %% 40. Inspect Mask Pixel Values
def inspect_mask_values(level, n=10):
    mask_dir = SEGMENTATION_ROOT / level / "Labels" / "Mask"
    masks = get_image_files(mask_dir)
    if len(masks) == 0:
        print("No masks found.")
        return pd.DataFrame()

    samples = random.sample(masks, min(n, len(masks)))
    results = []

    for mask_path in samples:
        try:
            mask = np.array(Image.open(mask_path))
            unique_values = np.unique(mask)
            results.append({
                "Mask": mask_path.name,
                "Shape": mask.shape,
                "Unique Values": unique_values[:30],
                "Number of Unique Values": len(unique_values)
            })
        except Exception as e:
            results.append({
                "Mask": mask_path.name,
                "Shape": None,
                "Unique Values": None,
                "Number of Unique Values": None
            })

    return pd.DataFrame(results)

mask_values_df = inspect_mask_values("Easy")
if not mask_values_df.empty:
    display(mask_values_df)

# %% 41. Check Whether Masks Are Binary
def check_binary_masks(level):
    mask_dir = SEGMENTATION_ROOT / level / "Labels" / "Mask"
    masks = get_image_files(mask_dir)
    binary_count, non_binary_count = 0, 0

    for mask_path in masks:
        try:
            mask = np.array(Image.open(mask_path))
            unique_values = np.unique(mask)
            if len(unique_values) <= 2:
                binary_count += 1
            else:
                non_binary_count += 1
        except Exception:
            pass

    return {
        "Difficulty": level,
        "Total Masks": len(masks),
        "Binary Masks": binary_count,
        "Non-Binary Masks": non_binary_count
    }

binary_results = [check_binary_masks(level) for level in difficulty_levels]
binary_df = pd.DataFrame(binary_results)
display(binary_df)

# %% 42. Overall Dataset Summary
total_classification = sum(classification_df["Images"]) if not classification_df.empty else 0
total_detection = len(detection_images)
total_segmentation = segmentation_df["Images"].sum() if not segmentation_df.empty else 0
total_masks = segmentation_df["Masks"].sum() if not segmentation_df.empty else 0

summary = pd.DataFrame({
    "Dataset": ["Classification", "Detection", "Segmentation Images", "Segmentation Masks"],
    "Count": [total_classification, total_detection, total_segmentation, total_masks]
})
display(summary)

# %% 43. Final Audit Report Printout
print("=" * 60)
print("CONCRETE DAMAGE DATASET AUDIT REPORT")
print("=" * 60)
print(f"\nTotal images found: {len(all_images):,}")
print(f"Corrupted images: {len(corrupted_df):,}")
print(f"Unique images: {unique_images:,}")
print(f"Duplicate copies: {duplicate_count:,}")
print(f"\nClassification images: {total_classification:,}")
print(f"Detection images: {total_detection:,}")
print(f"Segmentation images: {total_segmentation:,}")
print(f"Segmentation masks: {total_masks:,}")
print("\n" + "=" * 60)

# %% 44. Export Audit Results to CSV Files
folder_df.to_csv(OUTPUT_DIR / "folder_statistics.csv", index=False)
extension_df.to_csv(OUTPUT_DIR / "extension_statistics.csv", index=False)
classification_df.to_csv(OUTPUT_DIR / "classification_statistics.csv", index=False)
segmentation_df.to_csv(OUTPUT_DIR / "segmentation_statistics.csv", index=False)
if not dimensions_df.empty:
    dimensions_df.to_csv(OUTPUT_DIR / "image_dimensions.csv", index=False)

print("Audit results saved successfully to:")
print(OUTPUT_DIR)
```
